```python
import numpy as np
```

```python
arr = np.array([12,34,56,67,78,10,4,5,6])
```

```python
arr[arr>30]
```

```python
arr = np.array([
    [12,34,56,10],
    [90,56,20,10],
    [2,3,4,50]
])
```

```python
arr[arr>30]
```

```python
arr[(arr > 10) & (arr < 50)]
```

```python
arr[(arr > 30) | (arr%4 == 0)]
```
