```python
import numpy as np
```

```python
arr = np.arange(1,10)

print(arr)
print(arr[0])
print(arr[1])
print(arr[-1])
```

```python
print(arr[0:4])
print(arr[:4])
print(arr[4:])
print(arr[0::2])
print(arr[-1::-1])
print(arr[::2])
```

```python
arr = np.arange(1,21).reshape(4,5)

print(arr)
print(arr[0][1])
print(arr[-1][-1])
print(arr[0,1])
print(arr[-1,-1])
print(arr[1,2])
```

```python
print(arr)
print(arr[0:2])
print(arr[0:,1:4])
print(arr[:2, :2])
print(arr[-1::-1, -1::-1])
print(arr[:, -1])
print(arr[2:, 3:])
```
